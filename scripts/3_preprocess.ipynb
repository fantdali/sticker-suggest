```python
from PIL import Image
import imagehash
from tqdm import tqdm
import os

unique = {}
duplicates = []

DATA_DIR = '../data/sticker_suggest/'
def filepath(filename):
    return os.path.join(DATA_DIR, filename) 

image_paths = [
    filepath(f) for f in os.listdir(DATA_DIR)
]

for path in tqdm(image_paths):
    try:
        img = Image.open(path).convert("RGB")
        h = imagehash.phash(img)

        if h in unique:
            duplicates.append(path)
        else:
            unique[h] = path
    except:
        continue
```

```python
print(len(unique), len(duplicates))
```

```python
import matplotlib.pyplot as plt

for i, d in duplicates[:50]:
    plt.figure(figsize=(12, 4))
    plt.subplot(10, 5, 1)
    plt.imshow(Image.open(d))
    plt.title('Duplicate')
    plt.axis('off')
```

```python
for path in duplicates:
    os.remove(path)
```

```python
filter_paths = [filepath(f) for f in os.listdir(DATA_DIR)]
len(filter_paths)
```

```python
import numpy as np

old_paths = np.load("../data/paths.npy")
embeddings = np.load("../data/embeddings.npy")
```

```python
mask = np.array([p in filter_paths for p in old_paths])
print(mask.shape)
```

```python
filtered_paths_np = old_paths[mask]
filtered_embeddings_np = embeddings[mask]
```

```python
len(filtered_paths_np), filtered_embeddings_np.shape
```

```python
np.save("../data/paths_clean.npy", filtered_paths_np)
np.save("../data/embeddings_clean.npy", filtered_embeddings_np)
```
